# Conversión de DOCX a Markdown
Convierte todos los archivos `.docx` de `Documentos_prompts_base_docx` a `.md` en `Documentos_prompts_base_md`,
preservando la estructura de subcarpetas (p.ej. `1.Investigación`, `2.Descubrimiento`, etc.).

```python
# Instalar dependencias si no están disponibles
#%pip install mammoth markdownify --quiet
```

```python
import os
import mammoth
from markdownify import markdownify as md
from pathlib import Path

# ── Configuración ──────────────────────────────────────────────────────────────
BASE_DIR   = Path(".")                          # directorio raíz del proyecto
SRC_DIR    = BASE_DIR / "Documentos_prompts_base_docx"
DST_DIR    = BASE_DIR / "Documentos_prompts_base_md"
# ───────────────────────────────────────────────────────────────────────────────

converted  = []
errors     = []

# Recorrer todas las subcarpetas de la fuente
for docx_path in sorted(SRC_DIR.rglob("*.docx")):
    # Ruta relativa respecto a SRC_DIR (e.g. "1.Investigación/archivo.docx")
    rel_path = docx_path.relative_to(SRC_DIR)

    # Destino: misma estructura de subcarpeta, extensión cambiada a .md
    md_rel_path = rel_path.with_suffix(".md")
    md_path     = DST_DIR / md_rel_path

    # Crear la carpeta destino si no existe
    md_path.parent.mkdir(parents=True, exist_ok=True)

    try:
        # 1. mammoth convierte el .docx a HTML conservando estructura básica
        with open(docx_path, "rb") as f:
            result = mammoth.convert_to_html(f)
        html_content = result.value

        # 2. markdownify convierte el HTML a Markdown
        md_content = md(
            html_content,
            heading_style="ATX",   # encabezados con #
            bullets="-",            # listas con -
            strip=["a"],            # eliminar hipervínculos vacíos si los hay
        )

        # 3. Guardar el archivo .md
        md_path.write_text(md_content, encoding="utf-8")

        converted.append(str(md_rel_path))
        print(f"✅  {rel_path}  →  {md_rel_path}")

        # Mostrar advertencias de conversión de mammoth si las hay
        for msg in result.messages:
            print(f"   ⚠️  {msg}")

    except Exception as e:
        errors.append((str(rel_path), str(e)))
        print(f"❌  Error en {rel_path}: {e}")

# ── Resumen ────────────────────────────────────────────────────────────────────
print()
print("═" * 60)
print(f"Conversión completada: {len(converted)} archivo(s) convertido(s).")
if errors:
    print(f"Errores: {len(errors)}")
    for path, err in errors:
        print(f"  - {path}: {err}")
print("═" * 60)
```
